```python
#DATA PREPROCESSING
```

```python
#rename some columns
df.rename(columns={'bookId':'bookid', 'numRatings':'numofratings', 'likedPercent':'likedpercent',
                   'coverImg':'coverimg', 'bbeVotes':'bbevotes', 'ratingsByStars':'ratebystar', 'bbeScore':'bbescores'}, inplace=True)
```

```python
#remove the specific columns
df = df.drop(['firstPublishDate', 'setting', 'publishDate', 'characters', 'bookFormat', 'description'],
             axis=1)
```

```python
#detailed information of each column
df.info()
```

```python
#extract the first two words from the 'author' column
df['author'] = df['author'].str.split().str[:2].str.join(' ')
df['author'] = df['author'].str.replace(',', '')
df['author'] = df['author'].str.replace('.', '')
df['author'].head(10)
```

```python
#remove the word 'pages' and convert 'pages' column to int datatype
df['pages'] = df['pages'].astype(str).str.extract('(\d+)').fillna(0).astype('int32')
#convert 'price' column from object to float datatype
df['price'] = df['price'].astype(str).str.extract('(\d+)').astype('float')
#display the modified columns
df[['pages', 'price']]
```

```python
#keep only rows where 'language' is 'English'
df = df[df['language'] == 'English']
df['language']
```

```python
#include values having more than 0 pages
df = df[df['pages'] > 0]
```

```python
#drop the original 'ratingsbystar' column
df.drop('ratebystar', axis=1, inplace=True)
```

```python
#remove '[]' from the genres
df['genres'] = df['genres'].str.strip("[]")
df = df.drop(df[df['genres'] == ''].index[0])
```

```python
#keep only the first genre for each book
df['genres'] = df['genres'].str.split(',').str[0].str.strip().str.replace("'", "")
df['genres']
```

```python
#remove the rows having null/missing values.
df.dropna(axis=0, inplace=True)
#again check the missing values
df.isnull().sum()
```

```python
#check duplicate values of the dataset
df.duplicated().sum()
```

```python
def replace_outliers_with_median(series):
  if pd.api.types.is_numeric_dtype(series):
    median = series.median()
    lower_quartile = series.quantile(0.25)
    upper_quartile = series.quantile(0.75)
    iqr = upper_quartile - lower_quartile
    lower_bound = lower_quartile - 1.5 * iqr
    upper_bound = upper_quartile + 1.5 * iqr
    return series.apply(lambda x:median if x < lower_bound or x > upper_bound else x)
  else:
    return series

df_outlier_treated = df.apply(replace_outliers_with_median, axis=0)
```

```python
#ENCODING CATEGORICAL VARIABLES USING LABEL ENCODER
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
df['encoded_genres'] = label_encoder.fit_transform(df['genres'])
df[['encoded_genres', 'genres']].head(60)
```

```python
label_encoder = LabelEncoder()
df['encoded_author'] = label_encoder.fit_transform(df['author'])
df[['encoded_author', 'author']].head(60)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df[['encoded_genres', 'encoded_author', 'rating', 'bbescores', 'bbevotes', 'numofratings']]
y = df['likedpercent']
```

```python
#train-test split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=42)
```

```python
#variance of target variable
variance_likedpercent = df1['likedpercent'].var()
print(f'Variance of likedpercent: {variance_likedpercent:.2f}')
```

```python
#standardize numerical features
```

```python
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

ANN MODEL

```python
import tensorflow as tf
from tensorflow import keras
```

```python
model = keras.Sequential([
    keras.layers.Dense(64, activation='relu', input_shape=(6,)),
    keras.layers.Dense(32, activation='relu'),
    keras.layers.Dense(1)  #output layer for regression
])
```

```python
#compiling
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mean_squared_error'])
```

```python
#training
model.fit(X_train_scaled, y_train, epochs=60, batch_size=32, validation_split=0.1, verbose=1)
```

```python
#predict test set
y_pred = model.predict(X_test_scaled)
```

```python
#mean squared error and r-squared
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
```

```python
#print the result
print(f'Mean Squared Error: {mse:.2f}')
print(f'R-squared: {r2:.2f}')
```

RANDOM FOREST REGRESSOR & GRADIENT BOOSTING REGRESSOR

```python
# Example: Ensemble with RandomForestRegressor and GradientBoostingRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.ensemble import VotingRegressor

rf_model = RandomForestRegressor()
gb_model = GradientBoostingRegressor()

ensemble_model = VotingRegressor([('rf', rf_model), ('gb', gb_model)])
ensemble_model.fit(X_train, y_train)

ensemble_predictions = ensemble_model.predict(X_test)
ensemble_mse = mean_squared_error(y_test, ensemble_predictions)
print(f'Mean Squared Error (Ensemble): {ensemble_mse}')
```
